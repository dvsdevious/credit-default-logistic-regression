### General Preprocessing

The EDA identified several preprocessing requirements before modeling. Most variables contained valid values, however, undocumented categories were observed in the EDUCATION and MARRIAGE variables. In addition, categorical variables required encoding before they could be used by machine learning algorithms. Because the target variable is binary, it also needed to be separated from the predictor variables.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_excel('../data/raw/default_of_credit_card_clients.xls', header=1)
df.head()
```

```python
df_prep = df.copy()
```

Renaming the column

```python
df_prep = df_prep.rename(columns={'default payment next month':'DEFAULT'})
df_prep.head()
```

Dropping ID column

```python
df_prep = df_prep.drop(columns='ID')
df_prep.head()
```

#### EDUCATION

```python
edu_counts = df_prep['EDUCATION'].value_counts().sort_index()
edu_counts
```

```python
df_prep['EDUCATION'] = df_prep['EDUCATION'].replace([0,5,6,], 4)
```

```python
df_prep['EDUCATION'].value_counts().sort_index()
```

#### Marital Status

```python
mar_counts = df_prep['MARRIAGE'].value_counts().sort_index()
mar_counts
```

```python
df_prep['MARRIAGE'] = df_prep['MARRIAGE'].replace([0], 3)
```

```python
df_prep['MARRIAGE'].value_counts().sort_index()
```

Dummies

```python
df_prep = pd.get_dummies(df_prep, columns=['SEX', 'EDUCATION', 'MARRIAGE'], drop_first=True)
```

```python
df_prep
```

The categorical variables SEX, EDUCATION, and MARRIAGE were converted into binary indicator variables using one-hot encoding. The drop_first=True option was used to avoid multicollinearity by removing one reference category from each variable.

```python
df_prep.to_csv('../data/processed/preprocessed_credit_default.csv', index=False)
```
